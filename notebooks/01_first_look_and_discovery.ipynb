## 2. Discover a real truth yourself

The safest, most satisfying early wins are **things you find in the data** — un-leakable, and they *are* the core lesson. Run the three cells below. Each is ~10 lines of pandas and each overturns a common SEO belief.

Every number is **computed live from the shipped CSV** — nothing is hardcoded.

```python
import pandas as pd, numpy as np
df = pd.read_csv("data/raw/content_refresh_anonymized.csv")
print(df.shape[0], "rows,", df.shape[1], "columns")
df.head(3)
```

### Discovery A — "High search volume means more traffic." Does it?

```python
corr = df["search_volume"].corr(df["impressions_90d"])
print(f"Correlation between search_volume and impressions_90d: {corr:.3f}")
print("Near zero -> keyword search volume barely predicts the traffic a page actually gets.")
```

### Discovery B — the CTR cliff by position
Click-through rate is not flat: it collapses as you move down the results.

```python
visible = df[df["impressions_90d"] >= 100]
ctr_by_pos = visible.groupby("position_tier")["ctr"].mean().sort_values(ascending=False)
print(ctr_by_pos.round(4).to_string())
ctr_by_pos.plot(kind="bar", title="Mean CTR by position tier (impressions >= 100)", ylabel="CTR");
```

### Discovery C — is longer content the lever?
Compare word count for **declining** vs **growing** pages.

```python
wc = df.groupby("trend_direction")["word_count"].median()
print(wc.round(0).to_string())
print("\n'down' vs 'up' pages have almost the same median word count -> length is not the lever.")
```

Pick **one** of these and write a few lines below. This is your Week-1 discovery — you'll reference it in your Week-1 research-question write-up (on the InternHQ board).

- Redo Discovery A but only for pages with `impressions_90d > 0` — does the correlation change?

- Find another belief to test: does `content_age_days` relate to `trend_direction`? Does `avg_position` relate to `ctr`?

**Rules:** describe what you observe as *observed / directional* — never "I proved Google's algorithm." Keep client data out of anything you publish.

```python
# Your discovery here
# Belief tested: "Older pages are the ones that decline." Does content_age_days relate to trend_direction?

# Sanity check first: Discovery A's filter (impressions_90d > 0) removes nothing in this sample.
print("Rows with impressions_90d == 0:", int((df["impressions_90d"] == 0).sum()), "of", len(df))

# 1) Median page age by trend bucket
print("\nMedian content_age_days by trend_direction:")
print(df.groupby("trend_direction")["content_age_days"].median().round(0).to_string())

# 2) Share of pages that are 'down' within each age tier
tier_order = df.groupby("age_tier")["age_tier_order"].first().sort_values().index
down_rate = (df.assign(is_down=df["trend_direction"].eq("down"))
               .groupby("age_tier")["is_down"].mean().reindex(tier_order))
n_per_tier = df.groupby("age_tier").size().reindex(tier_order)
print("\nShare of pages trending 'down' by age tier:")
print(pd.DataFrame({"pages": n_per_tier, "down_rate": down_rate.round(3)}).to_string())
print(f"\nOverall 'down' rate: {df['trend_direction'].eq('down').mean():.3f}")

down_rate.plot(kind="bar", title="Share of pages trending down, by age tier", ylabel="down rate");
```

**What I observed (directional, on this 30,000-page anonymized sample):**

- I first tried the suggested redo of Discovery A with `impressions_90d > 0`, but it changes nothing: every page in this sample already has impressions (0 of 30,000 rows are zero). So that option is a no-op here, and the correlation stays ~0.001. Worth knowing before building on it.
- I tested a different belief instead: *"older pages are the ones that decline."* The data points the other way. The share of pages trending **down** falls steadily with age: about 67% for 31-90 day-old pages, 63% for 91-180, 52% for 181-365 and 43% for 365+ days (overall 54%). Median age also tells the same story: `down` pages are younger (216 days) than `stable` (300) or `up` (292) pages.
- Possible reading (a hypothesis, not a proof): newer pages may start with a short burst of visibility that fades, while older pages have already settled into a steady level. Pages in the 31-90 day tier are few (492), so I treat that group's rate with caution.
- This is an observed association in one sample. It does not show that age *causes* decline, and it says nothing about how Google's algorithm works.
